```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python
x1=np.array([1,2,3,4])
y1=np.array([2,3,4,6])
x2=np.array([6,3,4,7])
y2=np.array([2,5,7,9])
plt.plot(x1,y1,label='dataset1')
plt.plot(x2,y2,'--',label='dataset2')
plt.legend()
plt.xlabel("index")
plt.ylabel("value")
plt.title("multi-line")
plt.grid(True)
plt.show()
```

```python
#fill between area betwwen two line 
x=np.array([0,1,2,3])
y1=np.array([2,4,1,3])
y2=np.array([1,2,0,1])
plt.plot(x,y1,label='upper-line') 
plt.plot(x,y2,label='lower-line')
plt.fill_between(x,y1,y2,color='orange',alpha=0.3) 
plt.legend()
plt.show()
```

```python
#scatter plot 
x=np.array([1,2,3,5,10,23,12,30,40,55])
y=np.array([10,20,1,2,3,40,50,60,70,80])
plt.scatter(x,y) 
plt.show()
```

```python
#two scatter plot compare 
#group a 


x1=np.array([120,130,140,150,160,170])
y1=np.array([50,60,70,80,90,100])

#groupb 
x2=np.array([130,140,120,110,150,160])
y2=np.array([48,50,60,70,40,80]) 
plt.scatter(x1,y1,color='navy',label='group a') 
plt.scatter(x2,y2,color='orange',label='group b')
plt.legend()#compareing two grouo 
plt.show() 
```

```python
#bubble plot 

x=[1,2,3,4,5] 
y=[5,7,8,6,9] 
bubble_size=[60,200,300,100,400] 
plt.scatter(x,y,c='red', s=bubble_size,alpha=0.7,edgecolors='green',linewidths=2)
plt.show()
```

```python
#custom marker 
x=[3,5,7,9,11]
y=[8,12,14,10,16]
plt.scatter(x,y,marker='s',color='teal',s=120,alpha=0.6) 
plt.show()
```

```python
x=range(1,5) 
markers=['.','o','s','^'] 
linestyles=['-','--','-.',':'] 
for i ,(marker,linestyle) in enumerate(zip(markers,linestyles)): 
    plt.plot(x,[i]*4,marker=marker,linestyle=linestyle, 
             markersize=10,label=f"line{i+1}")
plt.legend()
plt.show()
```

```python
#uniform distrubation 
data=np.random.uniform(10,50,800) 
plt.hist(data,bins=20,color='orange',edgecolor='blue')
plt.show()
```

```python
data=np.random.normal(loc=10,scale=5,size=100) 
plt.hist(data,bins=20,density=True,color='violet',edgecolor='black')
plt.show()
```

```python
data1=np.random.normal(20,3,500) 
data2=np.random.normal(35,4,500) 
fig,axs=plt.subplots(1,2,figsize=(12,5))
axs[0].hist(data1,bins=18,color='green',edgecolor='black')
axs[0].set_title('dataset 1')
axs[1].hist(data2,bins=18,color='red',edgecolor='black') 
axs[1].set_title('dataset 2') 
for ax in axs:
    ax.set_xlabel('values')
    ax.set_ylabel('count')
plt.show()
```
