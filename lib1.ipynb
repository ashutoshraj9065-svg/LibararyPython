```python
import pandas as pd 
import numpy as np 
```

```python
#pandas Series
#series is a one dimensional data structure in python that can hold data of any type 

number=[10,20,30,40] 
s1=pd.Series(number)
print(s1)
```

```python
# series indexing 
print(s1[2])
```

```python
#series labeling 

number=[10,20,30,40] 
s1=pd.Series(number,index=['a','b','c','d'])
print(s1)
```

```python
#create series dictonary

data={"grades":[90,98,93],'attendence':[50,56,65]}
df=pd.DataFrame(data)
print(df)
```

```python
grades=('math':90,'science':70,'english':40)
s1=pd.Series(grades,index=['math','science','english'])
print(s1)
```
