```python
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns

data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales": [120000, 135000, 128000, 150000, 170000, 165000],
    "Profit": [20000, 25000, 22000, 30000, 38000, 35000],
    "Ad_Spend": [10000, 12000, 11000, 15000, 18000, 17000]
}

df = pd.DataFrame(data)
df
```

```python
sns.lineplot(data=df,x="Month",y="Sales")
plt.show()
```

```python
# sns.scatterplot(data=df,x="Ad_Spend",y="Sales") 
# plt.show()

sns.histplot(data=df,x="Sales",bins=5,kde=True) 
plt.show()
```
