```python
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python


data = {
    "Department": ["IT", "IT", "HR", "Sales", "IT", "HR", "Sales", "Sales"],
    "Salary": [50000, 60000, 40000, 45000, 70000, 50000, 55000, 65000]
}

df = pd.DataFrame(data)
df
```

```python
# sns.countplot(data=df,x="Department")
# plt.show()

sns.barplot(data=df,x="Department",y="Salary")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

fig, ax = plt.subplots(1, 2, figsize=(10, 4))

sns.countplot(
    data=df,
    x="Department",
    ax=ax[0]
)

sns.barplot(
    data=df,
    x="Department",
    y="Salary",
    ax=ax[1]
)

ax[0].set_title("Employee Count")
ax[1].set_title("Average Salary")

plt.tight_layout()
plt.show()
```
