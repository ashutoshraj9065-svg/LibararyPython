```python
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python
data = {
    "Ad_Spend": [10, 20, 30, 40, 50, 60],
    "Sales":    [100, 150, 210, 260, 320, 390]
}

df=pd.DataFrame(data) 
corrr=df["Ad_Spend"].corr(df["Sales"])
print(corrr) 
#matlab strong positive realtionship haii sirr 
# df
# sns.scatterplot(data=df,x="Ad_Spend",y="Sales")
```

```python
plt.scatter(df["Ad_Spend"],df["Sales"])
plt.xlabel("Ad_Spend")
plt.ylabel("Sales")
plt.title("Realtionship")

plt.show()
```
